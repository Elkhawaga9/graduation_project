```python
import os
import time
from pathlib import Path
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

# --- CONFIGURATION ---
CHROMA_PATH = Path("../vector_store/chroma_db")
MODEL_NAME = "all-MiniLM-L6-v2"
```

```python
# --- THE TEST EXAM ---
# These are the questions we will ask your "Brain" automatically
TEST_QUESTIONS = [
    "What is Amdahl's Law?",
    "Explain the difference between big endian and little endian.",
    "What is the purpose of the cache memory?",
    "What is a zombie process in Unix?",
    "How does virtual memory handling work?"
]
```

```python
# --- MAIN SCRIPT ---
print("🚀 Starting Automated Database Health Check...")
print(f"📂 Database Path: {CHROMA_PATH.resolve()}")

if not CHROMA_PATH.exists():
    print(f"❌ CRITICAL ERROR: Database not found at {CHROMA_PATH}")
    exit()
```

```python
# 1. Load DB
print("🧠 Loading Neural Network (Embeddings)...")
embedding_function = HuggingFaceEmbeddings(model_name=MODEL_NAME)
vector_db = Chroma(persist_directory=str(CHROMA_PATH), embedding_function=embedding_function)

print(f"✅ System Online. Total Knowledge Chunks: {vector_db._collection.count()}")
print("="*60)

```

```python
# 2. Run the Test Loop
for i, question in enumerate(TEST_QUESTIONS):
    print(f"\n🧪 TEST #{i+1}: {question}")
    
    start_time = time.time()
    
    # Search for top 1 best answer
    results = vector_db.similarity_search_with_score(question, k=1)
    
    end_time = time.time()
    duration = end_time - start_time
    
    if not results:
        print("❌ FAILED: No results found.")
        continue
        
    doc, score = results[0]
    
    # --- PRINT REPORT ---
    print(f"⏱️  Retrieval Time: {duration:.4f} seconds")
    print(f"🎯 Relevance Score: {score:.4f} (Lower is better)")
    print(f"📄 Source: {doc.metadata.get('source', 'Unknown')}")
    print(f"📌 Topic: {doc.metadata.get('topic', 'General')}")
    print("-" * 20)
    print(f"📖 RETRIEVED TEXT:\n{doc.page_content[:400]}...") # Show first 400 chars
    print("="*60)

print("\n✅ DIAGNOSTIC COMPLETE.")
```
